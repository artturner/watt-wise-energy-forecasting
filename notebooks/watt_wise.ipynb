```python
import pandas as pd
```

```python
df = pd.read_csv('../data/Energy_consumption.csv')
```

```python
df.dtypes
```

```python
# Convert to datetime and set as index for ARIMA/SARIMA)

df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Sort by Timestamp

df = df.sort_values('Timestamp')

# Drop duplicates

df = df.drop_duplicates(subset='Timestamp')

# Or check for duplicates to determine if there a need to drop)
# df['Timestamp'].duplicated().sum()

# Set as index

df.set_index('Timestamp', inplace=True)
```

```python
# Add datetime features

df['hour'] = df.index.hour
df['day'] = df.index.day
df['month'] = df.index.month
df['weekday'] = df.index.weekday
```

```python
# Categorical encoding
# drop_first=True (1)to avoid perfect multicollinearity (dummy variable trap), and (2) to reduce dimensionality.

df = pd.get_dummies(df, columns=['DayOfWeek', 'HVACUsage', 'LightingUsage'], drop_first=True)
df['Holiday'] = df['Holiday'].map({'Yes': 1, 'No': 0})
```
